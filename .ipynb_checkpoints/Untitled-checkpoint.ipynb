# Data Cleaning

```python
df_cleaned = df.copy()
```

```python
df_cleaned.head()
```

```python
df_cleaned['MSZoning'] = df_cleaned['MSZoning'].fillna(df_cleaned['MSZoning'].mode()[0])
```

```python
df_cleaned['Exterior1st'] = df_cleaned['Exterior1st'].fillna(df_cleaned['Exterior1st'].mode()[0])
```

```python
df_cleaned['BsmtFinSF2'] = df_cleaned['BsmtFinSF2'].fillna(0)
```

```python
df_cleaned['TotalBsmtSF'] = df_cleaned['TotalBsmtSF'].fillna(0)
```

```python
df_cleaned.isnull().sum()
```

```python
df_cleaned = df_cleaned.dropna(subset=['SalePrice'])
```
