```python
def plotting(col,num):
    plt.subplot(4,4,num)
    sns.histplot(df[col],kde=True)
    plt.title(col)
    

numeric_cols=["LotArea",
    "OverallCond",
    "YearBuilt",
    "YearRemodAdd",
    "BsmtFinSF2",
    "TotalBsmtSF",
    "SalePrice"]
plt.figure(figsize=(20, 20))
n=0
for col in numeric_cols:
    n+=1
    plotting(col,n)

plt.tight_layout() 
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(x='MSZoning',y='SalePrice',data=df)
plt.title("SalePrice By MSZoing")
plt.xlabel("MSZoning")
plt.ylabel("SalePrice")

plt.show()
```
